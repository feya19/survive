```python
# ================================================================
# UNIVERSAL ENTERTAINMENT ANALYTICS
# ADVERTISING PREDICTION — LIGHTGBM (REVISED)
# Kaggle Notebook: jalankan %pip di cell terpisah bila diperlukan:
# %pip -q install lightgbm joblib scikit-learn
# ================================================================

import glob
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from lightgbm import LGBMRegressor, early_stopping, log_evaluation
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.inspection import permutation_importance

warnings.filterwarnings("default")
SEED = 42
TARGET = "revenue"
MODEL_FILE = "lightgbm_advertising_revenue.joblib"

# 1. LOAD DATA
CSV_PATH = None  # Isi path manual jika ingin memilih file tertentu.
if CSV_PATH is None:
    csv_candidates = glob.glob("/kaggle/input/**/*.csv", recursive=True)
    if not csv_candidates:
        raise FileNotFoundError(
            "CSV tidak ditemukan. Tambahkan dataset lewat Kaggle Notebook > Add Data."
        )
    print("CSV yang ditemukan:")
    for i, path in enumerate(csv_candidates):
        print(f"{i}: {path}")
    CSV_INDEX = 0  # Ganti jika file dataset bukan pilihan pertama.
    CSV_PATH = csv_candidates[CSV_INDEX]

df = pd.read_csv(CSV_PATH)
print("File:", CSV_PATH, "| Shape awal:", df.shape)

# 2. STANDARDISASI NAMA KOLOM
df.columns = (
    df.columns.astype(str).str.strip().str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_")
)
required = [
    "date", "platform", "campaign_type", "industry", "country",
    "impressions", "clicks", "ctr", "cpc", "ad_spend",
    "conversions", "cpa", "revenue", "roas"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Kolom wajib tidak ditemukan: {missing}\nKolom tersedia: {df.columns.tolist()}")

# 3. EDA AWAL
print("\nShape:", df.shape)
print("\nTipe data:\n", df.dtypes)
display(df.head())
display(df.describe(include="all").T)
display(
    df.isna().sum().to_frame("missing_count")
    .assign(missing_percent=lambda x: (100 * x["missing_count"] / len(df)).round(2))
    .sort_values("missing_percent", ascending=False)
)
print("Duplikat persis:", df.duplicated().sum())
for col in ["platform", "campaign_type", "industry", "country"]:
    print(f"\nDistribusi {col}:")
    display(df[col].value_counts(dropna=False))

# 4. CLEANING
df["date"] = pd.to_datetime(df["date"], errors="coerce")
numeric_cols = [
    "impressions", "clicks", "ctr", "cpc", "ad_spend",
    "conversions", "cpa", "revenue", "roas"
]
for col in numeric_cols:
    if df[col].dtype == "object":
        df[col] = (
            df[col].astype(str).str.replace(",", "", regex=False)
            .str.replace("$", "", regex=False)
            .str.replace("₹", "", regex=False).str.strip()
        )
        df[col] = df[col].replace({"": np.nan, "nan": np.nan, "None": np.nan})
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df.loc[~np.isfinite(df[col]), col] = np.nan

cat_cols = ["platform", "campaign_type", "industry", "country"]
for col in cat_cols:
    df[col] = df[col].astype("string").str.strip()
    df[col] = df[col].replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})

df = df.drop_duplicates().dropna(subset=["date", TARGET]).copy()
# Asumsi revenue negatif tidak valid. Sesuaikan bila dataset merekam refund/net revenue.
df = df[df[TARGET] >= 0].copy()
# Spend negatif tidak valid untuk model ini.
df.loc[df["ad_spend"] < 0, "ad_spend"] = np.nan

print("\nShape setelah cleaning:", df.shape)
print("Rentang tanggal:", df["date"].min(), "s.d.", df["date"].max())
if len(df) < 30:
    raise ValueError("Data terlalu sedikit untuk temporal train/validation/test split.")

for col in ["impressions", "clicks", "conversions", "ad_spend"]:
    print(f"Nilai negatif {col}: {(df[col] < 0).sum()}")

# 5. AUDIT METRIK (diagnostik saja; tidak dipakai sebagai fitur)
def safe_ratio(a, b):
    return a / b.replace(0, np.nan)

df["ctr_recalculated"] = safe_ratio(df["clicks"], df["impressions"])
df["cpc_recalculated"] = safe_ratio(df["ad_spend"], df["clicks"])
df["cpa_recalculated"] = safe_ratio(df["ad_spend"], df["conversions"])
df["roas_recalculated"] = safe_ratio(df["revenue"], df["ad_spend"])

print("\nAudit korelasi metrik (bukan bukti definisi identik):")
for original, recalculated in [
    ("ctr", "ctr_recalculated"), ("cpc", "cpc_recalculated"),
    ("cpa", "cpa_recalculated"), ("roas", "roas_recalculated")
]:
    valid = df[[original, recalculated]].dropna()
    if len(valid) >= 2 and valid[original].nunique() > 1 and valid[recalculated].nunique() > 1:
        print(f"{original}: correlation={valid[original].corr(valid[recalculated]):.4f}")
    else:
        print(f"{original}: data valid tidak cukup")

# 6. EDA PLOT
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
df[TARGET].hist(bins=35, ax=axes[0])
axes[0].set_title("Revenue Distribution")
df["ad_spend"].hist(bins=35, ax=axes[1])
axes[1].set_title("Ad Spend Distribution")
df.groupby(df["date"].dt.to_period("M"))[TARGET].mean().plot(ax=axes[2], marker="o")
axes[2].set_title("Average Revenue per Month")
axes[2].tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()

display(df[numeric_cols].corr(numeric_only=True).round(3))
display(
    df.groupby("platform", dropna=False)[TARGET]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
)

# 7. FEATURE ENGINEERING
# Asumsi: ad_spend adalah anggaran yang diketahui ketika prediksi dibuat.
# Jika ad_spend merupakan realisasi akhir kampanye, hasil bukan pre-campaign forecast.
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day_of_week"] = df["date"].dt.dayofweek
df["day_of_month"] = df["date"].dt.day
df["quarter"] = df["date"].dt.quarter
df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
df["weekday_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["weekday_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)
df["log_ad_spend"] = np.log1p(df["ad_spend"])
df["target_log_revenue"] = np.log1p(df[TARGET])

numeric_features = [
    "ad_spend", "log_ad_spend", "year", "month", "day_of_week",
    "day_of_month", "quarter", "month_sin", "month_cos",
    "weekday_sin", "weekday_cos"
]
categorical_features = ["platform", "campaign_type", "industry", "country"]
feature_columns = numeric_features + categorical_features
df = df[np.isfinite(df["target_log_revenue"])].copy()

# 8. TEMPORAL SPLIT: 70% TRAIN, 15% VALIDATION, 15% TEST
df = df.sort_values("date").reset_index(drop=True)
n = len(df)
train_end, val_end = int(n * 0.70), int(n * 0.85)
train_df = df.iloc[:train_end].copy()
val_df = df.iloc[train_end:val_end].copy()
test_df = df.iloc[val_end:].copy()
if min(len(train_df), len(val_df), len(test_df)) == 0:
    raise ValueError("Salah satu split kosong.")
print("\nTEMPORAL SPLIT")
for name, part in [("TRAIN", train_df), ("VALIDATION", val_df), ("TEST", test_df)]:
    print(f"{name}: {len(part)} baris | {part.date.min().date()} - {part.date.max().date()}")

X_train, X_val, X_test = (
    train_df[feature_columns].copy(),
    val_df[feature_columns].copy(),
    test_df[feature_columns].copy()
)
y_train = train_df["target_log_revenue"].copy()
y_val = val_df["target_log_revenue"].copy()
y_test = test_df["target_log_revenue"].copy()

# 9. PREPROCESSING: fit hanya pada TRAIN
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median"))])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
train_arr = preprocessor.fit_transform(X_train)
val_arr = preprocessor.transform(X_val)
test_arr = preprocessor.transform(X_test)
feature_names = preprocessor.get_feature_names_out()

# Explicit DataFrames preserve feature names and avoid feature-name warnings.
def as_named_frame(array, index):
    if hasattr(array, "toarray"):
        array = array.toarray()
    return pd.DataFrame(array, columns=feature_names, index=index)

X_train_p = as_named_frame(train_arr, X_train.index)
X_val_p = as_named_frame(val_arr, X_val.index)
X_test_p = as_named_frame(test_arr, X_test.index)

# 10. TRAIN LIGHTGBM WITH REGULARIZATION + EARLY STOPPING
model = LGBMRegressor(
    objective="regression",
    n_estimators=1500,
    learning_rate=0.02,
    num_leaves=15,
    max_depth=6,
    min_child_samples=30,
    subsample=0.8,
    subsample_freq=1,
    colsample_bytree=0.8,
    reg_alpha=1.0,
    reg_lambda=5.0,
    random_state=SEED,
    n_jobs=-1,
    verbosity=-1
)
model.fit(
    X_train_p, y_train,
    eval_set=[(X_val_p, y_val)],
    eval_metric="rmse",
    callbacks=[
        early_stopping(stopping_rounds=100, verbose=False),
        log_evaluation(period=100)
    ]
)
print("Best iteration:", model.best_iteration_)

# 11. EVALUATION ON ORIGINAL REVENUE SCALE
def predict_revenue(X_processed):
    return np.maximum(np.expm1(model.predict(X_processed)), 0)

def evaluate_model(name, X_processed, y_log):
    actual = np.expm1(np.asarray(y_log))
    predicted = predict_revenue(X_processed)
    result = {
        "split": name,
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "R2": r2_score(actual, predicted)
    }
    print(
        f"\n{name}: MAE={result['MAE']:,.4f}, "
        f"RMSE={result['RMSE']:,.4f}, R2={result['R2']:.4f}"
    )
    return result

metrics_df = pd.DataFrame([
    evaluate_model("TRAIN", X_train_p, y_train),
    evaluate_model("VALIDATION", X_val_p, y_val),
    evaluate_model("TEST (final)", X_test_p, y_test)
])
display(metrics_df)

# 12. ACTUAL VS PREDICTED + TEST PREDICTIONS
test_actual = np.expm1(y_test.to_numpy())
test_pred = predict_revenue(X_test_p)
max_value = max(float(test_actual.max()), float(test_pred.max()))
plt.figure(figsize=(7, 6))
plt.scatter(test_actual, test_pred, alpha=0.65)
plt.plot([0, max_value], [0, max_value], linestyle="--")
plt.xlabel("Actual Revenue")
plt.ylabel("Predicted Revenue")
plt.title("Test Set: Actual vs Predicted Revenue")
plt.tight_layout()
plt.show()

test_predictions_df = pd.DataFrame({
    "date": test_df["date"].to_numpy(),
    "actual_revenue": test_actual,
    "predicted_revenue": test_pred
})
test_predictions_df["residual"] = (
    test_predictions_df["actual_revenue"] - test_predictions_df["predicted_revenue"]
)
test_predictions_df.to_csv("advertising_test_predictions.csv", index=False)

# 13. FEATURE IMPORTANCE
importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance_gain": model.booster_.feature_importance(importance_type="gain"),
    "importance_split": model.booster_.feature_importance(importance_type="split")
}).sort_values("importance_gain", ascending=False)
print("\nTop 20 gain importance:")
display(importance_df.head(20))

top_features = importance_df.head(20).sort_values("importance_gain")
plt.figure(figsize=(10, 7))
plt.barh(top_features["feature"], top_features["importance_gain"])
plt.xlabel("Gain Importance")
plt.title("Top 20 LightGBM Feature Importance")
plt.tight_layout()
plt.show()

# Permutation importance memakai named DataFrame (menghindari warning nama fitur).
perm = permutation_importance(
    model, X_test_p, y_test,
    scoring="neg_mean_squared_error",
    n_repeats=5, random_state=SEED, n_jobs=-1
)
permutation_df = pd.DataFrame({
    "feature": X_test_p.columns,
    "permutation_importance_mean": perm.importances_mean,
    "permutation_importance_std": perm.importances_std
}).sort_values("permutation_importance_mean", ascending=False)
print("\nTop 20 permutation importance:")
display(permutation_df.head(20))

# 14. MONTE CARLO-READY VALIDATION RESIDUALS
# Residual validation = actual revenue - predicted revenue pada skala asli.
# Gunakan residual validation untuk sampel error simulasi; jangan residual train.
val_actual = np.expm1(y_val.to_numpy())
val_pred = predict_revenue(X_val_p)
validation_residuals_df = pd.DataFrame({
    "date": val_df["date"].to_numpy(),
    "actual_revenue": val_actual,
    "predicted_revenue": val_pred,
    "residual": val_actual - val_pred
})
validation_residuals_df.to_csv("advertising_validation_residuals.csv", index=False)
print("\nRingkasan residual validation untuk Monte Carlo:")
display(validation_residuals_df["residual"].describe())

# 15. LEAKAGE AUDIT
excluded_features = [
    "impressions", "clicks", "ctr", "cpc",
    "conversions", "cpa", "roas", "revenue"
]
print("\nFitur yang dikecualikan:", excluded_features)
print("Fitur yang dipakai:", feature_columns)
print(
    "PERHATIAN: pastikan semua fitur benar-benar tersedia saat waktu prediksi. "
    "Temporal split tidak otomatis mencegah leakage dari agregasi masa depan."
)

# 16. SAVE MODEL BUNDLE + CSV OUTPUTS
model_bundle = {
    "model_type": "LightGBMRegressor",
    "target": TARGET,
    "target_transform": "log1p",
    "feature_columns": feature_columns,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "preprocessor": preprocessor,
    "feature_names_after_preprocessing": list(feature_names),
    "model": model,
    "best_iteration": int(model.best_iteration_ or model.n_estimators),
    "metrics": metrics_df.to_dict(orient="records"),
    "training_date_min": str(train_df["date"].min()),
    "training_date_max": str(train_df["date"].max()),
    "validation_date_min": str(val_df["date"].min()),
    "validation_date_max": str(val_df["date"].max()),
    "test_date_min": str(test_df["date"].min()),
    "test_date_max": str(test_df["date"].max()),
    "excluded_features": excluded_features,
    "residuals_file": "advertising_validation_residuals.csv",
    "assumption_note": (
        "Interpretasikan sebagai pre-campaign forecast hanya jika ad_spend "
        "adalah budget yang diketahui saat prediksi; jika realisasi, model "
        "memprediksi revenue bersyarat pada spend aktual."
    )
}
joblib.dump(model_bundle, MODEL_FILE)
importance_df.to_csv("advertising_feature_importance.csv", index=False)
permutation_df.to_csv("advertising_permutation_importance.csv", index=False)
metrics_df.to_csv("advertising_evaluation_metrics.csv", index=False)

print("\n========== SELESAI ==========")
for output_file in [
    MODEL_FILE,
    "advertising_feature_importance.csv",
    "advertising_permutation_importance.csv",
    "advertising_evaluation_metrics.csv",
    "advertising_test_predictions.csv",
    "advertising_validation_residuals.csv"
]:
    print("-", output_file)

# 17. CONTOH INFERENCE (opsional; sesuaikan nilainya)
# new_campaign = pd.DataFrame([{
#     "ad_spend": 1000.0, "log_ad_spend": np.log1p(1000.0),
#     "year": 2026, "month": 11, "day_of_week": 2,
#     "day_of_month": 15, "quarter": 4,
#     "month_sin": np.sin(2*np.pi*11/12), "month_cos": np.cos(2*np.pi*11/12),
#     "weekday_sin": np.sin(2*np.pi*2/7), "weekday_cos": np.cos(2*np.pi*2/7),
#     "platform": "Google Ads", "campaign_type": "Search",
#     "industry": "E-commerce", "country": "Indonesia"
# }])
# transformed = preprocessor.transform(new_campaign[feature_columns])
# transformed = as_named_frame(transformed, new_campaign.index)
# print("Predicted revenue:", float(predict_revenue(transformed)[0]))
```
